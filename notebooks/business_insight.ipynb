```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('cleaned_data.csv')
df_filtered = df.rename(columns={'Cluster_ID': 'กลุ่มลูกค้า'})
custom_colors = ['#1d5ac4', '#21cdb6', '#8f8f8f']

SLIDE_W = 900
SLIDE_H = 500

# ---------------------------------------------------------
# กราฟที่ 1: Donut Chart
# ---------------------------------------------------------
main_cats = ['โซเชียลมีเดีย', 'แพลตฟอร์มวิดีโอ', 'โทรทัศน์', 'สื่อ ณ จุดขาย', 'แอปพลิเคชั่นแชท/ส่งเมสเสจ']
cat_counts = {cat: df['S_Influence'].str.contains(cat).sum() for cat in main_cats}

other_cats = ['เว็บไซต์', 'บิลบอร์ด/ป้ายโฆษณา', 'รถเมล์/รถไฟฟ้า']
other_sum = sum([df['S_Influence'].str.contains(cat).sum() for cat in other_cats])
cat_counts['อื่นๆ'] = other_sum

df_pie = pd.DataFrame(list(cat_counts.items()), columns=['Platform', 'Count'])

fig1 = px.pie(
    df_pie, values='Count', names='Platform',
    title='สัดส่วนช่องทางที่มีอิทธิพลต่อการตัดสินใจซื้อ',
    hole=0.4, color_discrete_sequence=custom_colors
)
fig1.update_traces(textinfo='percent+label', textfont_size=11)
fig1.update_layout(
    width=SLIDE_W, height=SLIDE_H,
    title_font=dict(size=16, family="Tahoma", color="#333"),
    title_x=0.5,
    legend=dict(orientation="h", yanchor="bottom", y=-0.25, xanchor="center", x=0.5),
    margin=dict(t=50, b=80, l=40, r=40),
    paper_bgcolor='white', plot_bgcolor='white'
)
fig1.show()
```

```python
# ---------------------------------------------------------
# กราฟที่ 2: Horizontal Bar Chart
# ---------------------------------------------------------
all_categories = main_cats + other_cats
counts = {cat: df['S_Influence'].str.contains(cat).sum() for cat in all_categories}
df_influence = pd.DataFrame(list(counts.items()), columns=['Platform', 'Count']).sort_values(by='Count', ascending=True)

fig2 = px.bar(
    df_influence, x='Count', y='Platform',
    title='จำนวนผู้ได้รับอิทธิพลจากโฆษณาจำแนกตามช่องทาง',
    orientation='h', text_auto=True,
    color='Count', color_continuous_scale='Blues'
)
fig2.update_layout(
    width=SLIDE_W, height=SLIDE_H,
    title_font=dict(size=16, family="Tahoma", color="#333"),
    title_x=0.5,
    xaxis_title="จำนวนผู้ใช้งาน (คน)",
    yaxis_title="ช่องทางสื่อโฆษณา",
    margin=dict(t=50, b=60, l=160, r=40),
    paper_bgcolor='white', plot_bgcolor='white'
)
fig2.show()
```

```python
# ---------------------------------------------------------
# กราฟที่ 3: S_Occasion แบ่งตาม Cluster
# ---------------------------------------------------------
fig3 = px.histogram(
    df_filtered, x='S_Occasion', color='กลุ่มลูกค้า',
    title='แพลตฟอร์มที่ใช้งานประจำ แบ่งตามกลุ่มลูกค้า',
    barmode='group', color_discrete_sequence=custom_colors,
    template='plotly_white'
)
fig3.update_layout(
    width=SLIDE_W, height=SLIDE_H,
    title_font=dict(size=16, family="Tahoma", color="#333"),
    title_x=0.5,
    xaxis_title="แพลตฟอร์มที่ใช้งานประจำ",
    yaxis_title="จำนวนลูกค้า (คน)",
    legend=dict(orientation="h", yanchor="bottom", y=-0.3, xanchor="center", x=0.5),
    margin=dict(t=50, b=100, l=60, r=40),
    paper_bgcolor='white', plot_bgcolor='white'
)
fig3.show()
```

```python
# ---------------------------------------------------------
# กราฟที่ 4: Festival Behavior
# ---------------------------------------------------------
fig4 = px.histogram(
    df_filtered, y='S_Usage(festival)', color='กลุ่มลูกค้า',
    title='พฤติกรรมการใช้สื่อช่วงเทศกาลสงกรานต์',
    barmode='group', color_discrete_sequence=custom_colors,
    template='plotly_white'
)
fig4.update_layout(
    width=SLIDE_W, height=SLIDE_H,
    title_font=dict(size=16, family="Tahoma", color="#333"),
    title_x=0.5,
    xaxis_title="จำนวนลูกค้า (คน)",
    yaxis_title="พฤติกรรมช่วงเทศกาล",
    legend=dict(orientation="h", yanchor="bottom", y=-0.3, xanchor="center", x=0.5),
    margin=dict(t=50, b=100, l=180, r=40),
    paper_bgcolor='white', plot_bgcolor='white'
)
fig4.show()
```
